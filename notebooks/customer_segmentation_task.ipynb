```python
import sys
from pathlib import Path

sys.path.append(str(Path("../src").resolve()))

from customer_segmentation.aggregation import aggregate_data
from customer_segmentation.cleaning import clean_sign_up_date, filter_canada_purchases, remove_consecutive_duplicates, rename_columns

from customer_segmentation.export import export_and_display_results
from customer_segmentation.features import compute_loyalty_score
from customer_segmentation.load import load_data
```
